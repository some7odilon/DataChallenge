```python
import pandas as pd

tables = pd.read_excel('dataChallenge.xlsx', sheet_name=None, header=None)

def ligne_texte(ligne):
    cellules = [str(c).strip() for c in ligne if pd.notna(c) and str(c).strip() != '']
    return ' | '.join(cellules)[:150]

for nom, df in tables.items():
    print(f"\n=== {nom} {df.shape} ===")
    for i in range(min(6, len(df))):
        print(ligne_texte(df.iloc[i]))

print("\n--- Feuilles qui mentionnent l'achèvement ---")
for nom, df in tables.items():
    trouve = any(
        'chèvement' in str(c).lower() or 'achevement' in str(c).lower()
        for c in df.to_numpy().ravel() if pd.notna(c)
    )
    if trouve:
        print(nom)
```
